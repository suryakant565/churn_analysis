```python
import pandas as pd 
import numpy as np
```

```python
df_customer = pd.read_parquet('../cleaned_data/df_customer_cleaned.parquet')
df_support = pd.read_parquet('../cleaned_data/df_support_cleaned.parquet')
df_subs = pd.read_parquet('../cleaned_data/df_subscription_cleaned.parquet')
```

```python
# create a new col using existing col - churn flag
# where customer is churned if cancellation_date is not null or mentioned
# merge df
# cheack the shape of data is correct 
# save the merged data for further data analysis
```

```python
# create a new col using existing col - churn flag
# where customer is churned if cancellation_date is not null or mentioned

df_subs['churn_flag'] = np.where(df_subs['cancellation_date'].notna(),1,0)
```

```python
df_subs.head(2)
```

```python
# merge df

df = (df_subs
    .merge(df_customer, on = 'customer_id', how='left')
    .merge(df_support, on = 'customer_id', how='left'))
```

```python
# cheack the shape of data is correct 
df.shape
```

```python
# save the merged data for further data analysis
df.to_parquet('../merged_data/merged_churn_data.parquet', index=False)
```
